### Corpus, lexicón y bolsa de palabras

**Corpus**: Un conjunto de documentos en texto plano. (páginas web, noticias, tweets, publicaciones,...). Es básicamente todo el conjunto de datos límpio sin analizar. 

**Lexicón**: El conjunto completo de palabras DISTINTAS que aparecen en el corpus, para analizar las frecuencias teniendo encuenta la ley de zipf. 

**Bolsa de palabras**: 
* El orden no importa: solo cuántas veces aparece cada una. 
* Dimensiones ...
* Es una representación o visulazación que ayuda a visualizar de manera descriptiva de que se esta hablando en un corpus, la tésis. Y es fundamental apra el procesamiento de lenguaje natural(????) que se usa para representar documentos textuales de tal manera que una máquina la pueda entender. 

### Matriz término-documentos
Es una forma de representar la palabras del texto como una tabla de números: filas=documentos, columnas=términos. 

Las filas representan las unidades de estudio (los documentos) de texto que se analizarán, y las columnas de la matriz representan los términos de l texto que se utilizarán en el análisis. 

### Stemming y lematización
Es lo último que se puede usar para reducción de dimensionalidad. 
#### Steemming: 
Cortar una palabra a una raíz heurística, aunque alguans veces no puede servir ya que a la raíz que se lleve puede no tener significado, aunque es muy rápido. 
#### Lematización. 
Directamente usa la categoría y gramática para hallar el lema, que es la raíz válida de las palabras que se están estudiando y son bastante específicas del idioma.

Ej: las palabras, correr, corriendo, corren stemming puede llevar a l razí "corr" que no tiene mucho sentido de una maner bastante rápida, pero lematización puede llevarme al lema "correr". 

Ambas técnicas ayudan al proceso de reducción de dimensiones de la matriz.

#### Pesar términos y compara documentos

Un texto se vuelve en datos que es contar frecuencias de palabras gracias a las bolsas de palabras. Ahora, queremos darlesuna importancia a las palabras de forma que podamos comparar documentos. 

Vamos a ver formar de caracterizar documentos y ver como se pueden comparar, que tan parecidos son y que se puede hacer eso. 

#### **TF-IDF**
Es un separador del peso de las palabras que intenta resolver el problema y darle mayor peso a las palabras que en un documento pero que sean buenas. 

**corpus: conjunto de documentos**

**TF (frecuencia de términos)***: Importancia local de documentos, mide que tan seguido aparece un término $t$ - en un documento $d$. Es la frecuencia cruda de cada una de los términos. 

También se suele presentar como:

*Conteo crudo: frecuencia directa, cruda. 

*Una frecuencia relativa: como frecuencias/total de palabras. 

*Log-normalizada: 1 + log(conteo) para amortiguar conteos simples ya que muchas veces una alta frecuencia no es importancia. 



**IDF(Frecuencia inversa de documento
)**: Lo que hace es ver que tan rara es una palabra dentro de todo el CORPUS, pues, una palabra rara dentro del corpus permite difernciar un texto de otro pues las palabras raras son muy informativas

$$idf(t) = log(\frac{N}{df(t)})$$

Donde $N = \text{número total de documentos}$ y $df(t) = \text{documentos que contienen t}$. Las pocas palabras son menos informaticas. IDF las penaliza de forma natural. 

**TF-IDF: el producto**

$$tfidf(t, d) = tf(t, d) x idf(t)$$

Alto cuando el término es frecuente en este documento pero raro en el corpus: cuando es característico de ese documentos. 


**Documentos como vectores**

## Atención en 3 pasos. 

1. comparar: Cada palabra mide cuánto se relaciona con cada otra palabra de la frase. 
El modelo o mecanismo de atención hace estas mediciones de las relaciones con las palabras que se tienen. 

**Ejemplo: "Fui al banco a sacar dinero"**

fui - no. 
al - no.
banco - si.
a - no

dinero - si.

2. Ponderar: Esas relaciones se convierten en pesos que suman 1: a quién prestar más o menos atención. 

Luego, la ponderación de cada palabra nos permite decidir a cuales de esos términos podemos dar más atención.
