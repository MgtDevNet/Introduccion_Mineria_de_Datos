Contexto del ejercicio

En este laboratorio trabajaremos con la base de datos DataEjemploSQL.csv, que contiene información simulada de un proceso productivo en una empresa manufacturera. Cada registro corresponde a una observación asociada a una fábrica, una sección y una línea de producción, junto con la variable TiempoPNP, que representa el tiempo de parada no programada.

El tiempo de parada no programada corresponde al tiempo en que una línea o proceso se detiene de manera inesperada, es decir, por causas no previstas dentro de la operación normal. Este tipo de paradas puede estar relacionado con fallas técnicas, problemas operativos, falta de insumos u otras interrupciones no planeadas. Por esta razón, TiempoPNP puede interpretarse como un indicador de ineficiencia o de posibles problemas en el proceso productivo.

A partir de esta base de datos, el objetivo del ejercicio es utilizar SQL para consultar, filtrar, ordenar y resumir información, de manera que podamos responder preguntas básicas sobre el comportamiento del proceso. Más allá de practicar sintaxis, la idea es usar SQL como una herramienta para explorar y entender los datos, identificando qué líneas presentan mayores tiempos de parada, qué grupos concentran más observaciones y cómo se comporta esta variable en distintos niveles de agregación.

### librería sqlite3 dentro de python. 

sqlite3 es una herramiento interna de python que básicamente es **un motor de bases de datos SQL completo que no requiere un servidor**.

A diferencia de postgresql y mysql no se necesita software aparte, configurar usuarios y encender un servicio. Vive directamente en el código y guarda un archvio único en el disco duro. 

Es algo serverless, ya que tenemos los datos pero no necesitamos un servidor, se conecta directamente a un archivo .db.

Toda la base de datos es un solo archivo y usa la misma sintaxis que los motores grandes. 

Es muy útil para análisis de datos personas, aplicaciones de escritorio, scripts sencillos y prototipos de machine learning.

### Ejemplo donde se crea desde cero una tabla en el sistema de bases de datos.

```python
# Conectarse a la base de datos (si no existe, se creará)

#Esto viene a ser como crear un servidor virtual de base de datos,
# que se va a guardar en un archivo llamado mi_base_de_datos.db
conn = sqlite3.connect('../../datos/mi_base_de_datos.db')

# Crear un cursor para ejecutar consultas
cursor = conn.cursor()

# Ejecutar una consulta SQL para crear una tabla
cursor.execute('''CREATE TABLE IF NOT EXISTS usuarios
                (id INTEGER PRIMARY KEY, nombre TEXT, edad INTEGER)''')

# Insertar datos en la tabla
cursor.execute("INSERT INTO usuarios (nombre, edad) VALUES (?, ?)", ('Juan', 30))
cursor.execute("INSERT INTO usuarios (nombre, edad) VALUES (?, ?)", ('María', 25))

# Ejecutar una consulta SQL para seleccionar datos
cursor.execute("SELECT * FROM usuarios")

# Obtener los resultados de la consulta. 
#Los resultados se guardarn EN UNA LISTA y para obtenerlos
#hay que instanciar una variable cursor.fetchall
filas = cursor.fetchall()

# Imprimir los resultados
for fila in filas:
    print(fila)

# Cerrar el cursor y la conexión
cursor.close()
conn.close()

#Es necesario cerrar las conexiones para liberar memoria,
#no dejar que se bloqueen los archivos y que los datos 
#estén seguros con respecto a que no se editen luego de 
#haber hecho las ediciones o manipulaciones.

#ESTE CÓDIGO  de cursor.execute es para cualqier tipo de linea de sql
```

```python
##Qué pasa si cambiamos INNER por LEFT? por RIGHT?
consulta = """
SELECT t1.linea, t1.min,t3.min_fecha
FROM t3
LEFT JOIN t1 ON t1.Linea = t3.Linea
"""

# Cargar los resultados del JOIN en un DataFrame
pd.read_sql_query(consulta, conn)
```

```python
consulta="""select t1.linea, t1.min
from t1
union
select t2.linea, t2.min_fecha
from t2"""
pd.read_sql_query(consulta, conn)
```

```python
##Ahora veamos qué hace UNION
consulta = """
SELECT t1.linea, t1.min,t3.min_fecha
FROM t1
LEFT JOIN t3 ON t1.Linea = t3.Linea
UNION
SELECT t1.linea, t1.min,t2.min_fecha
FROM t1
LEFT JOIN t2 ON t1.Linea = t2.Linea
"""

# Cargar los resultados del JOIN en un DataFrame
pd.read_sql_query(consulta, conn)
```

```python
##Calculando el promedio por año y Línea:
consulta="""
select Linea,
año,
(select avg(TiempoPNP)
from tabla2 as a
where a.linea=b.linea and a.año=b.año) as promedio
from tabla2 as b
group by linea, año
order by linea, año
"""
pd.read_sql_query(consulta, conn)
```

```python
##Corrigiendo los años:
consulta="""
select linea,
TiempoPNP,
FechaProgramacion,
CASE
when length(FechaProgramacion)=10 then substr(FechaProgramacion,7,4)
else substr(FechaProgramacion,6,4)
END as año
from tabla1
where  FechaProgramacion>'2018-09-11' and linea in ('Linea Baby 4','Linea Baby 7')
"""
tabla2=pd.read_sql_query(consulta, conn)
tabla2.to_sql('tabla2',if_exists='replace',index=False,con=conn)
```

```python
##Corrigiendo el promedio
consulta="""
select Linea,
año,
(select avg(TiempoPNP)
from tabla2 as a
where a.linea=b.linea and a.año=b.año) as promedio
from tabla2 as b
group by linea, año
order by linea, año
"""
pd.read_sql_query(consulta, conn)
```

```python
#ROW_NUMBER() OVER (ORDER BY Linea) AS numero_fila,
#SUM
#AVG
consulta="""
select Linea,
año,
(select AVG(TiempoPNP)
from tabla2 as a
where a.linea=b.linea and a.año=b.año) as promedio
from tabla2 as b
group by linea, año
order by linea, año
"""
pd.read_sql_query(consulta, conn)
```
