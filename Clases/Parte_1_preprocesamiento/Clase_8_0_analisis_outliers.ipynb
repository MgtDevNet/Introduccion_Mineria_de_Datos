### Recordemos algunas definiciones básicas de tipos de datos: 

Los datos se pueden ver como un objeto, un vector de características. Es decir, si pensamos en una tabla de una base de datos relacional entonces un registro será un vector de características con información de una persona por ejemplo. Cuando se ve la tabla por medio de columas se entienden como variables o atributos.

Por tanto, un dato se puede pensar como un objeto que tiene su vector de características que es como el registro con la información de una persona y que tiene sus atributos que vienen a ser las variables

Ahora bien, esos atributos o variables se asocian a un tipo de datos ya sea numérico o categórico(cualitativo). 

Variables cualitativas:

* **Ordinal**: Se puede ordenar como el estrato, rango militar. 

* **Nominal**:categorías, estados o nombres de cosas como el color de pelo o estado civil.

* **Binario**: Atributo nominal con solo 2 estados. 
    * **Binario simétrico**: Ambos resultados son igual de importantes como el sexo. 
    * **Binario asimétrico**: los resultados no son igualmente importantes como el resultado de un exámen médico.

        Nota: Por convención siempre se toma 1 como el resultado más importante, por ejemplo, VIH positivo. 

Variables Cuantitativas: discretas o continuas. 

* **Intervalo**: Mide una escala de iguales unidades pero que no hay un cero absoluto como lo es la temperatura donde el 0° no implica ausencia de temperatura. 

* **Razón**: Hay un cero que representa ausencia. 


### Repaso de descripciones estadísticas básicas. 

* Motivación: 
    * Tener una vista general de los datos. 

* Medidas de tendencia central: 
    * Media, mediana, moda, rango medio. 
    * media, media ponderada, media recortada(media pero con un recorte pequeño, aproximadamente el 2% de los datos atípicos para que esta no se vea tan influenciada por estos datos)

* Medidas de dispersión o variabilidad: 
    * ¿Cómo se distribuyen los datos?
    * Rangos, cuartiles y rango intercuartil. 
    * Desviación estándar y varianza. 
    * 5 números mágicos, boxplot, ...

* Análisis gráficos
    * Histogramas, gráficos de dispersión, QQ, etc. 

* Simetría y sesgo: 
Los datos en la mayoría de aplicaciones reales no son simétricos por ello mismo 
hay medidas para medir que tan simétrica puede ser una distribución o medir directamente el sesgo. 

    * Sesgo positivo: moda es menor a la mediana. 
    * Sesgo negativo: moda es mayor a la mediana. 
    
    Donde la moda es donde hay mayor frecuencia. 

* Medidas de Dispersión: 
    * Cuartiles, outliers y boxplots: 
    - Cuartiles Q1(percentil 25), Q3(peqrcentil 75). 
    - IQR: Q3 - Q1. 
    - Resúmen 5 números: min, Q1, median, Q3, max. 
    - Boxplot, donde los extremos de las cajas son los cuartiles y los bigotes son 1.5 el IQR. 
    - Se considera outlier aquel valor afuera de estos bigotes.

    Sirve para comparar poblaciones que estén en las mismas unidades, mas no para decir que hay mucha dispersión o no. 

    * Varianza y desviación estandar (obviamente las estimaciones puntuales por máxima verosimilitud)

# Análisis de Outliers(Datos atípicos)

Por definción, un Outlier es una observación que se desvía tanto de las otras observaciones como para despertar sospechas que fue generado por un mecanismo diferente. Es decir, cuando se tiene un conjunto de datos, se espera que que estos se hayan dado por un mecanismo aleatoria el cual puede estudiarse por medio de su distribución, pero, si hay datos que parecen que fueron dados por un mecanismo alatorio diferente entonces los datos serán más raros. 

Por tanto, dependiente del mecanismo es que hay diferentes formas para tratarlo. 

Hay tres tipos de datos atípicos: 

* **Dato atípico global**: Parece que los datos siguen una distribución particular, por eso siguen una tendencia, pero hay uno que para nada sigue esa tendencia y por tanto puede ser que se sea porque no siga la misma distribución y por ende se sale de la dispersión. O tal vez que en verdad haya un dato muy raro 

* ***Dato atípico contextual o condicional***: Hay que entender el contexto en el que se da para definir que en verdad sea un dato atípico. Por ejemplo, notese la imagen de abjao de temperatura, si se hace un boxplot no mostraría un datos atípico. Pero, si lo analizamos de una manera contextual teniendo en cuenta el comportamiento de la temperatura durante es año, es notable que a medida que se va a mitad de año la temperatura crece, por tanto, que junio tenga una temperautra como diciembre el algo demasiado raro, lo cual según el contexto si lo debemos considerar un dato atípico así sea que un boxplot no lo vea como tal. 


* **Dato atípico colectivo**: Esta más relacionado a cosas inusuales en cuanto a dispersión de los datos. Por ejemplo, en un ataque cierbénitico donde de un momento a otro empiezen muchas transacciones, más de las esperadas, muchas más.

## Acercamiento univariado
 - Son valores atipicos en una sola variable. 
 - Son buenos para detección de extremos en una sola dimensión. 
 - Lo malo es que si se ve desde una perspectiva diferente, por ejemplo en más dimensiones, puede que en verdad no sea un data extremo. 

### Métodos univariados: 
 - IQR con el boxplot. 
 - Z-score y Z-score modificado. 
 - Identificación de valores extremos a partir de 1, 2 o 3 desvíos de la media. 

## Acercamiento multivariado
 - Los valores atípicos multivariados se pueden encontrar en un espacio n-dimensional. Es decir, como en un conjunto de varias variables. 
 - Para esto es necesario ajustar un modelo. 

### Métodos multivariados
 - Análisis globales: Clustering usando medidas de distancia como Mahalanobis. 

 - Local outlier Factor (LOF)
    - Método de detección de outliers basado en distancias. 
    - Calcula un score de outlier  a partir de una distancia que se normaliza por densidad. 

 - Métodos basados en árboles de búsqueda: IsolationForest.
