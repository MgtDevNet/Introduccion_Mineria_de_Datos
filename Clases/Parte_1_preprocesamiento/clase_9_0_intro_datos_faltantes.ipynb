# --- Datos Faltantes ---
* Origen. 
* Tratamientos: 
    * Eliminación.

Los datos faltantes cuando se trabajan con metodologías de machine learning tiene un efecto bastante significativo con respecto a las métricas de desempeño de un modelo al igual que pasa con los datos atípicos. Por tanto, nace la necesidad de tratarlos y tener un buen desempeño en nuestros modelos. 

* El problema de los datos faltantes está presente en el análisid e datos desde los orígenes del almacenamiento. 

* Este problema de valores faltantes se vuelve incluso un dilema propio del proceso KDD (descubrimiento de conocimiento) donde los volúmenes de datos se incrementan y con ellos la probabilidad de inconsistencias y faltantes. 

## Algunas razones de inconsistencias
* Factores propios del procedimiento: Formularios mal diseñados, errores de programación, etc.  

* Negativa a responder. 

* Respuestas inaplicables. 

Por tanto, muchas veces un dato vacío no necesariamente es un error y tal vez tratarla puede generar sesgos que no existían orginalmente en los datos. Por ello también una muestra piloto puede ser imporante para predecir posibles fallos y prever esto. 

Los datos faltantes dificultan un análisis, un manejo inapropiado de los mismos pueden resultar en conclusiones engañosas y pueden limitar la generalización del conocimiento alcanzado. 

## Mecanismos de datos faltantes
Algunos de los mecanismos que generan los datos faltantes son: 

* 1. Outliers tratados como datos faltantes: Cuando se conocen los límites de las diferentes variables del dataset, los datos que caen fuera del rango definido se deben etiquetar como faltantes. Es decir, la imputación del datos aítipico es volverlo faltante. 

* 2. Missing At Random **MAR**: 
    + La proba de que la variable Y tenga un dato faltante depende de X, pero no de Y. 
    + El patrón de datos faltantes se puede predecir partir de variables del conjunto de datos. 
    ejemplo: que haya un dato vacío en la edad ya que depende del género. 

    Es decir, su valor faltante se puede explicar con otra variable. 

* 3. Missing completely at random **MCAR**: 
    * La probabilidad de que la variable Y tenga un dato faltante es independiente de X. 
    * Los datos existente en Y son una muestra al azar de los valores de Y.

    Es decir, no hay una razón o dependiencia aparente para explicar el valor vacío.  

* 4. Non-ignorable missing data (Missing not at random **MNAR**): 
    * El valor de la variable que falta está relacionado con la razón por la que falta. 
    ejemplo: una pregunta esta vacía porque la pregunta es tal vez "fuerte" y no se quiere responder.

    Es decir, la razón del valor vacío reside en la complejidad de la pregunta.  


## Tratamientos de datos faltantes

- **Utilizar solo registros completos**: Esto va a depender de cómo es el origen del faltante. Se recomienda que se use solo en los casos en que el mecanismo de faltante sea MCAR, completamente aleatorio. Es decir, los vacíos no tienen una razón aparente. 

- **Borrar casos seleccionados o variables**: La eliminación de registrros ante la presencia de falantes puede utilizarse cuando hay un patrón **no aleatorio** de los datos faltantes. Ojo, si la eliminación disminuye la utilidad de los datos entonces no es tan efectiva.

- **Imputación de datos**: métodos para rellenar la información que falta. 

- **Aproximaciones basadas en modelado**: Se basa en imputaciones múltiples.  


## Imputación de datos. 

- Método de relleno.
- Imputar es el proceso de estimar datos faltantes de una observación a partir de valores válidos de otras variables. 
- **Hay que tener precaución ya que pueden generar sesgos en los datos reales y los imputados**

- 1. **Sustitución de casos**. Se reemplaza con valores no observador. Debería ser realizado por un experto en esos datos. 

- 2. **Sustitución de Medias**. Se reemplaza utilizando el promedio calculado de los valores presentes. Debe verificarse que los datos ajusten a una distribución normal, si los datos están sesgados es mejor utilizar la mediana. 

Desventajas: La varianza estimada de la nueva variable no es válida porque está atenuada por los valores repetidos. Además de distrosión de la distribución y que las correlaciones estarán atenuadas debido a la replicación de un solo valor constante. 

- 3. **Cold Deck** (no es lo más usual pero puede ser útil)
* Selecciona valores o usa relaciones obtenidas de fuentes distintas de la base de datos actual. 
* Se sustituye un valor constante derivado de fuentes externas o de investigación previa.
* Tiene la misma desventaja de aplicar la media, además de que llenar con otra base de datos es dificíl hallar una base de datos con las características que necesito.

- 4. **Hot Deck**
Se reemplazan los datos faltantes con valores obtenidos de registros que son lo más similares (lo que apliqué en PLE que se basa en buscar los registros que se aparezcan más y con eso llenar esos valores faltantes).j

Ventajas:
* Es bastante simple.
* Conserva los niveles de medición y pues son llenados bastante realistas.
* Finaliza el proceso de imputación con un conjunto de datos completo. 

Desventajas: 
* La dificultad de definir que es similar y que no lo és. 

![hot deck ejemplo](../imagenes/hot_deck.png)

Notese que como el registro 3 tiene la mayor cantidad de similitudes con el registro 2 entonces por ello es que se llenan con el valor del registro 2. 

- 5. **Regresión**
* Se usa el análisis de regresión para predecir valores faltantes a partir de variables relacionadas en el conjutno de datos. 

* Se pueden usar regresiones simples o múltiples. 

* Se identifican variables independiente y dependiente. 

Es una buena estrategia para rellenar valores de variables continuas.

Sin embargo, solo es buena cuando hay datos vaciós en una sola variable, pues las otras se necesitan para el entrenamiento del modelo de regresión.

Por tanto, si hay muchos datos faltantes en diferentes variables pues esto ya no será efectivo.

- 6. **Imputación de datos múltiple (MICE)**

Es una técnica que opera bajo el supuesto de que el origen de los datos faltantes es Missging At Random (MAR), es decir, la probabilidad de que un valor falte Y se puede explicar dado X pero no Y.

Proceso
![proceso de MICE](../imagenes/mice.png)

1. Se hace imputación simple (A TODAS LAS VARIABLES). 
2. Se toma 1 de las variable y se vuleve a poner en NA, como estaba originalmente antes del paso 1. 
3. Ajustar un modelo como en regresión que permita a partir de las variables ya imputadas por el paso 1 que prediga los valores faltantes de la variable del paso 2. 
4. Reemplazar los valores faltantes por los predichos del paso anterior. 

Luego, repertir esto con cada una de las variables. La idea sería realizar esto almenos 10 veces para que la metodología converja a valores más razonables. 

![ejemplo mice](../imagenes/mice_ejemplo.png)

Ventajas: 
* No produce sesgo (depende del modelo)
* Puede ser usado en cualqueir tipo de análisis. 
*Fácil de usar.

Desventajas:
* Hay que pensar en un modelo adecuado. 
* Puede ser costoso computacionalmente. 
* Genera un dataset completo por cada iteración.
