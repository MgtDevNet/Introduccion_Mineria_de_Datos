## Acercamiento univariado
 - Son valores atipicos en una sola variable. 
 - Son buenos para detección de extremos en una sola dimensión. 
 - Lo malo es que si se ve desde una perspectiva diferente, por ejemplo en más dimensiones, puede que en verdad no sea un data extremo. 

### Métodos univariados: 
 - IQR con el boxplot. 
 - Z-score y Z-score modificado. 
 - Identificación de valores extremos a partir de 1, 2 o 3 desvíos de la media. 

--------------

 - Mediana
 - IQR
 - Generalmente regla de decisión +- 1.5*IQR

## 2. Z-score
Z-score es una métrica que indica cuántas desviaciones estándar tienene una observación de la media muestral, asumiendo una distribución gaussiana.

Por tanto, es necesario que **la distribución de los datos sea a lo menos simétricas, más no necesariamente normal**, sino pues los datos atípicos pueden ser erróneos. 





Es básicamente una estandarización de los datos llevandolo a una N(0,1) y por tanto, si una observación es mayor a 3 o menor a -3 entonces de ahí podemos decir si un valor es aítipico o no. Por ello, el Z-score es una medida para ver a cuántas desviaciones estandar esta un datos de la media ya sea por encima o por debajo

$$z = \frac{x - \mu}{\sigma}$$

Limitaciones o desventajas es que es necesario que la distribución de los datos sea a lo menos simétrica. 

## 3. Z-score Modificado
En este caso, esta metología es robusta a que los datos no necesariamente tienen que ser normales o simétricos pues la media y la desviación pueden verse afectados por este sesgo de los datos dados por valores extremos. 

Por tanto, para distribuciones asimetricas este método es el indicado para revisar a cuantas desviaciones estandar esta cada dato de la media. 

La diferencia al Z-score normal es que:

* En lugar de media se usa mediana. 

* En lugar de desviación estandar se usa una medida de variabilidad llamada MAD (Desviación absoluta de la mediana).

    lo que hace mucho más estable y confiable.  

$$ Z_{mod}= M_i = \frac{0.6745(x_i - \tilde{x})}{MAD} $$

$$ MAD = \text{median}\{|x_i - \tilde{x}|\} $$

Donde $\tilde{x}$ es la MEDIANA de mi conjunto de datos y los $x_i$ los datos. 

MAD: es la desviaicón absoluta de la mediana. Se calcula obteniendo las desviaciones absolutas de cada dato respecto a la  mediana, y luego buscando la mediana de esos nuevos valores absolutos, como se ve en la formula de arriba. 

0.6745: Es una constante de escala basada en la normal para asegurar que el MAD sea consistente con la desviación estandar en datos sin valores aítpicos. 

Nota: como regla general si un datos es mayor 3.5 o menor a -3.5 entonces se considera atipico. 

![z_score_modificado](../imagenes/z_score_mod.png)

como puede verse en la imágen anterior los valores en azul son los datos que están a más de 3.5 según el Z-score modificado y pueden considerarse atípicos. 

Es muy útil el z-score modificado pues usa la mediana como medida principal de dispersión y la mediana no es afectada por los datos atípicos, lo cual si pasa con la media y por eso el z-score no es tan bueno cuando las distribuciones son sesgadas.

En síntesis: Z-score y Z-score modificado es ún método para identificación de outliers por medio de la estandarización de los datos, si la distribución es simétrica entonces se usa el Z-score, en caso contrario se usa el Z-score modificado. Pero, si se usa el Z-score modificado en una distribución simétrica tendrá el mismo resultado que el Z-score normal. 

**Nota: El z-score y el z-score modificado son método directamente diseñados para la detección de datos atípicos, pero operan bajo supuestos metodológicos diferentes. El z-score mide a cuantas desviaciones estandar esta cada dato respecto a la media pero es muy sensible a los mismos datos aptípicos y el z-score modificado busca arreglar este problema sustituyendo los parámetros sensibles usando la mediana y la desviación absoluta de la mediana(MAD) pero el criterio es parecido**
