 # MECANISMO DE ATENCIÓN

De manera intuitiva: cada palabra de la frase mira a todas las demás y presta atención para darle una representación a la palabra según el contexto según sus palabras a compañantes. 

## Atención en 3 pasos. 

1. comparar: Cada palabra mide cuánto se relaciona con cada otra palabra de la frase. 
El modelo o mecanismo de atención hace estas mediciones de las relaciones con las palabras que se tienen. 

**Ejemplo: "Fui al banco a sacar dinero"**

fui - no. 
al - no.
banco - si.
a - no
sacar - masomenos.
dinero - si.

analizando la palabra "banco" se relaciona con cada palabra: Fui, al, banco, a , sacar, dinero. 

2. Ponderar: Esas relaciones se convierten en pesos que suman 1: a quién prestar más o menos atención. 

Luego, la ponderación de cada palabra nos permite decidir a cuales de esos términos podemos dar más atención. 

ejemplo de pesos (solo una idea vaga)
fui - 0.025. 
al - 0.02.
banco - .
a - 0.005
sacar - 0.15.
dinero - 0.8.

El algoritmo se llama softmax. 

Esto de desarrolla con el producto punto. 

3. Combinar: Cada palabra se vuelve un promedio ponderado de las demás: su versión contextual. 

En este paso el modelo no solamente usa la info para saber cuales modelos son importantes si no también para crear otro vector. 

Al inicio "banco" representa un vector, luego de la combinación "banco" representa otro vector con pesos dado el contexto, es decir, la combinación de pesos que describe a la palabra banco.
