El muestreo Bootstrap es una técnica estadística que consiste en tomar múltiples submuestras con reemplazo a partir de una única muestra de datos original.

Su propósito fundamental es estimar la variabilidad o la distribución de un parámetro (como la media, la mediana o los coeficientes de un modelo) sin necesidad de recolectar nuevos datos en el mundo real.

1. Selección al azar: Eliges un elemento de tu muestra original de forma totalmente aleatoria.

2. Reemplazo: Registras ese valor y lo devuelves a la muestra original antes de la siguiente selección.

3. Repetición: Repites los pasos 1 y 2 exactamente $N$ veces hasta formar una muestra bootstrap.

4. Cálculo: Calculas el estadístico de interés (por ejemplo, la media) sobre esta nueva muestra.

5. Iteración: Repites los pasos 1 al 4 un número elevado de veces ($B = 1000$ o más) para construir la distribución empírica de tu estadístico.

Propiedad matemática clave: Al seleccionar $N$ elementos con reemplazo de un conjunto de $N$ datos, la probabilidad de que un elemento individual de la muestra original NO sea elegido en una muestra bootstrap es de aproximadamente $(1 - 1/N)^N \approx 1/e \approx 36.8\%$. Es por esto que cerca del $36.8\%$ de los datos quedan fuera en cada árbol de un Random Forest (los datos Out-of-Bag).

```python
import numpy as np

# Datos originales (N = 6)
datos_originales = np.array([165, 170, 175, 180, 185, 210])
N = len(datos_originales)

# Semilla para reproducibilidad
np.random.seed(42)

print("--- MUESTRA ORIGINAL ---")
print("Datos:", datos_originales)
print("Media original:", np.mean(datos_originales).round(2))
print("\n--- GENERANDO 3 MUESTRAS BOOTSTRAP ---")

# Generar 3 muestras bootstrap como demostración
medias_bootstrap = []
for i in range(1, 4):
    # np.random.choice con replace=True realiza el muestreo con reemplazo
    muestra_bs = np.random.choice(datos_originales, size=N, replace=True) # tomar los 6 datos aleatorios con reemplazo
    media_bs = np.mean(muestra_bs) #calcular media
    medias_bootstrap.append(media_bs) #agregar a la lista
    
    print(f"Muestra Bootstrap #{i}: {muestra_bs} | Media: {media_bs:.2f}")

# Simulación masiva (B = 1000 iteraciones) para crear toda una distribución
B = 1000
medias_masivas = [
    np.mean(np.random.choice(datos_originales, size=N, replace=True))
    for _ in range(B)
]

# Intervalo de confianza del 95% empírico
ic_inferior = np.percentile(medias_masivas, 2.5)
ic_superior = np.percentile(medias_masivas, 97.5)

print("\n--- RESULTADO TRAS 1000 REPETICIONES ---")
print(f"Intervalo de confianza del 95% para la media: [{ic_inferior:.2f}, {ic_superior:.2f}]")
```
