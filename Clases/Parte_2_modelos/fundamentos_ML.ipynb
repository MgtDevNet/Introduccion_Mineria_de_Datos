## FUNDAMENTOS DEL APRENDIZAJE AUTOMÁTICO

#### 1. ¿ Por qué ML?
- Por los avances en la capacidad computacional y al tener grandes volúmenes de datos nace la necesidad de encontrar relaciones entre los grandes volumenes de datos. Y por ende, encontrar algoritmos para poder hallar estas relaciones que nosotros no somos capaces de identificar por medio de estadística descriptiva. 

Acá ya tenemos modelos no paramétricos, por lo que dejamos de preocuparnos por los supuestos de la estadística clásica para poder tener un modelo de regresión. 

Además de la necesidad de la predicción o clasificación de datos. 

## Tipo de aprendizaje automático

### 1. Supervisado: 
- Aprende de los datos etiquetados. 
- Cada registros tiene su respuesta correcta. 
- Regresión: Predice un número continuo. 
- Clasificación: Predice una categoría. 

En estos tipos de modelos, decir que son datos etiquetados, es que tenemos la variable respuesta explícitamente para luego revisar 
si esta bien o no. 

### 2. No supervisado. 
- El algoritmo aprende de datos sin etiquetar.
- El algoritmo encuentra una estructura intrínseca por si solo. 
- Clustering: Agrupar grupos similares. 
- Reducción de dimensionalidad. (PCA, t-SNE)

Estos tipo de algoritmos no supervisados no tienen una variable respuesta de la cual se pueda revisar 
como esta la respuesta, sino que el modelo por si solo halla las relaciones intrínsecas.

## El Ciclo de ML
#### Representar -> Entrenar -> Evaluar -> Refinar

### 1. Datos. 
Recolección, limpieza, exploración, reducción de dimensionalidad, imputación, etc. 

### 2. Representar
Viene la parte de ingeniería de variables, normalizar, codificar y esta relacionado con el punto anterior. 

### 3. Entrenar
Elegir modelo de hiperparámetros con los datos de prueba. 

**Nota: Nunca se usa el 100% de los datos ya que hay que usar una parte para evaluar la capacidad predictiva del modelo. Por tanto, antes de realizar el modelo hay que separar los datos de entrenamiento y los de prueba. Obviamente, la idea es que el modelo use datos que nunca ha usado.**

### 4. Evaluar. 
Revisar métricas y diagnósticos. 

### 5. Refinar
Cambiar modelos, hiperparámetros, nuevas variables, más datos, etc. Todo con el objetivo de hallar el mejor modelo según el caso.

# Modelos, métricas y validación. 

¿Qué hace un buen modelo?
 - Un modelo que funciona bien en entrenamiento no garantiza nada. 
 - Lo que imporat es la generalzación: predecir bien datos nuevos por más raros que sean. 

Por tanto, nace la necesidad de tener métricas calculadas sobre el conjunto de prueba, tales como estadísticos de: 

    - ECM
    - Sesgo
    - Varianza

MSE -> varianza del error de predicción.
Sesgo -> underfitting sistemático. 

Y hay que mantener un equilibrio entre varianza y sesgo. Todo estimador tiene varianza y sesgo y un modelo es un estimador por lo que deben tener un equilibrio.
